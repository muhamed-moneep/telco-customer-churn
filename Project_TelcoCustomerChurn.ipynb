**Detecting Outliers**

> *Using IQR Method*



```python
def detect_outliers_iqr(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]
    print(f"{column} → Q1={Q1}, Q3={Q3}, IQR={IQR}")
    print(f"Lower={lower_bound}, Upper={upper_bound}")
    print(f"Number of outliers = {len(outliers)}\n")
    return outliers


outliers_Tenure = detect_outliers_iqr(df, 'tenure')
outliers_MonthlyCharges = detect_outliers_iqr(df, 'MonthlyCharges')
outliers_TotalCharges = detect_outliers_iqr(df, 'TotalCharges')
```

## 7. Feature Engineering

```python
# Group customers by subscription length (tenure)
df['tenure_group'] = pd.cut(df['tenure'], bins=[-1,12,24,48,72],
                             labels=['0-12','12-24','24-48','48-72'])

# Actual average monthly spend
df['avg_monthly_spend'] = df['TotalCharges'] / (df['tenure'].replace(0,1))

# Count of extra services the customer is subscribed to
extra_services = ['OnlineSecurity','OnlineBackup','DeviceProtection',
                   'TechSupport','StreamingTV','StreamingMovies']
df['num_extra_services'] = (df[extra_services] == 'Yes').sum(axis=1)

# Whether the customer has internet at all
df['has_internet'] = (df['InternetService'] != 'No').astype(int)

# Month-to-month contract (higher churn risk)
df['is_month_to_month'] = (df['Contract'] == 'Month-to-month').astype(int)

df[['tenure_group','avg_monthly_spend','num_extra_services','has_internet','is_month_to_month']].head()
```

**Explanation:** - `tenure_group`: grouping customers by subscription length helps us check whether new customers (0-12 months) churn more often than long-term ones.
- `avg_monthly_spend`: shows if a customer's actual spending differs from their stated monthly rate, which can flag sudden behavior changes.
- `num_extra_services`: the more add-on services a customer has (Security, Backup...), the more "locked in" they usually are, and the less likely to churn.
- `has_internet` and `is_month_to_month`: simple binary flags that summarize important information (contract type and internet access) in a numeric form the model can use directly.

## Probability & Hypothesis Testing

**Probability of Churn given Contract Type**

> * *Calculating P(Churn = Yes) and P(Churn = Yes | Contract type) using conditional probability*

```python
# Overall probability of churn (already computed earlier, recomputed here for clarity)
p_churn = (df['Churn'] == 'Yes').mean()
print(f"P(Churn = Yes) = {p_churn:.4f}")
print()

# Conditional probability: P(Churn = Yes | Contract = c) for each contract type
contract_probs = {}
for c in sorted(df['Contract'].unique()):
    subset = df[df['Contract'] == c]
    p_churn_given_c = (subset['Churn'] == 'Yes').mean()
    contract_probs[c] = p_churn_given_c
    print(f"P(Churn = Yes | Contract = {c}) = {p_churn_given_c:.4f}")

contract_probs = pd.Series(contract_probs).sort_values(ascending=False)

plt.figure(figsize=(7,5))
sns.barplot(x=contract_probs.index, y=contract_probs.values, palette='mako')
plt.axhline(p_churn, color='red', linestyle='--', label=f'Overall P(Churn) = {p_churn:.2f}')
plt.ylabel("P(Churn = Yes | Contract)")
plt.title("Conditional Probability of Churn by Contract Type")
plt.legend()
plt.show()
```

**Explanation:** Customers on a **Month-to-month** contract churn at a much higher conditional probability than the overall churn rate, while **One year** and **Two year** customers churn far less often. This confirms contract type is one of the strongest churn signals in the dataset.

**Bayes' Theorem — reversing the conditional probability**

> * *Given a customer churned, what is the probability they were on a Month-to-month contract?*

```python
# Bayes' theorem: P(Contract = Month-to-month | Churn = Yes)
#                = P(Churn = Yes | Contract = Month-to-month) * P(Contract = Month-to-month) / P(Churn = Yes)

p_mtm = (df['Contract'] == 'Month-to-month').mean()
p_churn_given_mtm = contract_probs['Month-to-month']

p_mtm_given_churn = (p_churn_given_mtm * p_mtm) / p_churn

print(f"P(Contract = Month-to-month)                = {p_mtm:.4f}")
print(f"P(Churn = Yes | Contract = Month-to-month)   = {p_churn_given_mtm:.4f}")
print(f"P(Churn = Yes)                                = {p_churn:.4f}")
print(f"\nP(Contract = Month-to-month | Churn = Yes)   = {p_mtm_given_churn:.4f}  (via Bayes' theorem)")

# Sanity check against the direct empirical calculation
direct_check = (df[df['Churn']=='Yes']['Contract'] == 'Month-to-month').mean()
print(f"Direct empirical calculation                 = {direct_check:.4f}")
```

**Explanation:** Bayes' theorem lets us flip the question around: instead of asking *"if a customer is Month-to-month, how likely are they to churn?"* we ask *"given that a customer churned, how likely is it they were Month-to-month?"*. The two numbers match the direct empirical calculation, confirming the vast majority of churners come from the Month-to-month segment.

**Hypothesis Testing — Effect of Payment Method on Churn**

> * *H0 (Null): Payment Method and Churn are independent (payment method has no effect on churn).*
> * *H1 (Alternative): Payment Method and Churn are NOT independent (payment method affects churn).*
> * *Test used: Chi-square test of independence (both variables are categorical), α = 0.05*

```python
contingency_payment = pd.crosstab(df['PaymentMethod'], df['Churn'])
print("Contingency table (PaymentMethod x Churn):")
print(contingency_payment)

chi2_pay, p_pay, dof_pay, expected_pay = chi2_contingency(contingency_payment)

print(f"\nChi-square statistic = {chi2_pay:.4f}")
print(f"Degrees of freedom    = {dof_pay}")
print(f"p-value               = {p_pay:.6f}")

alpha = 0.05
if p_pay < alpha:
    print(f"\np-value ({p_pay:.6f}) < alpha ({alpha}) -> Reject H0")
    print("Conclusion: There IS a statistically significant association between Payment Method and Churn.")
else:
    print(f"\np-value ({p_pay:.6f}) >= alpha ({alpha}) -> Fail to reject H0")
    print("Conclusion: No statistically significant association between Payment Method and Churn.")
```

**Explanation:** The p-value is far below 0.05, so we reject H0 — Payment Method is **not** independent of Churn. Looking at the contingency table, customers paying by **Electronic check** churn at a noticeably higher rate than customers using automatic bank transfer, automatic credit card, or mailed check.

**Bonus Hypothesis Test — Effect of Contract Type on Churn**

> * *H0: Contract type and Churn are independent*
> * *H1: Contract type and Churn are NOT independent*
> * *Test used: Chi-square test of independence, α = 0.05*

```python
contingency_contract = pd.crosstab(df['Contract'], df['Churn'])
print("Contingency table (Contract x Churn):")
print(contingency_contract)

chi2_c, p_c, dof_c, expected_c = chi2_contingency(contingency_contract)

print(f"\nChi-square statistic = {chi2_c:.4f}")
print(f"p-value               = {p_c:.6f}")

if p_c < alpha:
    print(f"\np-value ({p_c:.6f}) < alpha ({alpha}) -> Reject H0")
    print("Conclusion: Contract type has a statistically significant effect on Churn.")
else:
    print(f"\np-value ({p_c:.6f}) >= alpha ({alpha}) -> Fail to reject H0")
```

**Explanation:** This confirms numerically what we already saw in the Probability section: Contract type is very strongly associated with Churn (extremely small p-value), reinforcing it as one of the top retention levers available to the business.

**Explanation:** Even with just 2 components, churned customers (lighter color) visibly cluster more toward one side of the plot rather than being scattered randomly — a good sign that the reduced space still carries meaningful churn-related structure.

### 14.3 Choosing the number of clusters (Elbow Method)

```python
inertias = []
K_range = range(2, 8)
for k in K_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    km.fit(X_pca_2)
    inertias.append(km.inertia_)

plt.figure(figsize=(8,5))
plt.plot(list(K_range), inertias, marker='o')
plt.xlabel("Number of Clusters (k)")
plt.ylabel("Inertia (Within-Cluster Sum of Squares)")
plt.title("Elbow Method for Optimal k")
plt.tight_layout()
plt.show()
```

### 14.4 Clustering customers with K-Means

```python
k_final = 3
kmeans = KMeans(n_clusters=k_final, random_state=42, n_init=10)
clusters = kmeans.fit_predict(X_pca_2)

sil_score = silhouette_score(X_pca_2, clusters)
print(f"Silhouette Score for k={k_final}: {sil_score:.4f}")

plt.figure(figsize=(8,6))
plt.scatter(X_pca_2[:,0], X_pca_2[:,1], c=clusters, cmap='viridis', alpha=0.6, s=15)
plt.scatter(kmeans.cluster_centers_[:,0], kmeans.cluster_centers_[:,1],
            c='red', marker='X', s=200, label='Centroids')
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title(f"Customer Segments (K-Means, k={k_final}, on PCA components)")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
cluster_df = pd.DataFrame({'Cluster': clusters, 'Churn': y_train.values})
churn_by_cluster = cluster_df.groupby('Cluster')['Churn'].agg(['mean','count'])
churn_by_cluster.columns = ['Churn Rate', 'Number of Customers']
print("Churn rate by customer segment (cluster):")
print(churn_by_cluster)

plt.figure(figsize=(7,5))
sns.barplot(x=churn_by_cluster.index, y=churn_by_cluster['Churn Rate'], palette='mako')
plt.axhline(p_churn, color='red', linestyle='--', label=f'Overall churn rate = {p_churn:.2f}')
plt.xlabel("Cluster")
plt.ylabel("Churn Rate")
plt.title("Churn Rate by Customer Segment")
plt.legend()
plt.tight_layout()
plt.show()
```

**Explanation:** K-Means groups customers purely by the structure in the PCA-reduced feature space, with no knowledge of the Churn label. The fact that the resulting clusters still show clearly different churn rates confirms these segments are meaningful — one segment behaves like the classic "high risk" profile (short tenure, month-to-month, fiber optic, electronic check), while another looks like the "loyal" profile (long tenure, longer contracts, more add-on services). This segmentation can directly feed a retention strategy: target marketing and discounts at the high-risk segment first.

## 15. Conclusion

**Feature Selection** — the features that consistently came out important **across most methods** (Mutual Info, ANOVA, RFE, Lasso, Random Forest):

- `is_month_to_month`
- `tenure`
- `Contract_Two year`
- `TotalCharges`
- `InternetService_Fiber optic`
- `PaymentMethod_Electronic check`

**Probability** — P(Churn) is much higher for Month-to-month customers than the overall average, and Bayes' theorem confirms that the large majority of churners come from that segment.

**Hypothesis Testing** — Chi-square tests reject the null hypothesis for both Payment Method and Contract type (p-value < 0.05 in both cases): customers paying by **Electronic check**, and customers on **Month-to-month** contracts, churn at significantly higher rates than the rest of the customer base.

**PCA & Clustering** — reducing the 38 encoded features with PCA and segmenting customers with K-Means produced clusters with clearly different churn rates, confirming there is a distinguishable "high-risk" customer profile (short tenure, month-to-month contract, fiber optic internet, electronic check payment) versus a "loyal" profile (long tenure, longer contracts, more add-on services).

### Retention Recommendations
1. **Target Month-to-month, Electronic-check customers first** — this is the highest-risk segment identified by every method used (probability, hypothesis testing, feature selection, and clustering).
2. **Incentivize longer contracts** — offer a discount or perk for upgrading from Month-to-month to One/Two-year contracts, since contract length is the single strongest churn predictor.
3. **Encourage other payment methods** — nudge Electronic-check customers toward automatic bank transfer or credit card payment (e.g. small discount for autopay), since this payment method is significantly associated with churn.
4. **Bundle add-on services** — customers with more extra services (Security, Backup, Tech Support, Streaming) churn less; bundling these into cheaper packages could improve retention for at-risk customers.
5. **Use the customer segments for targeted campaigns** — the high-risk cluster from the PCA/K-Means analysis is a ready-made target list for a proactive retention campaign (special offers, check-in calls, etc.) instead of blasting the entire customer base.
